# 03 Embeddings

This notebook builds quarter-specific GraphSAGE embeddings and exports a pooled dataset for later regression experiments in a separate notebook.

```python
from pathlib import Path
import sys

PROJECT_ROOT = Path.cwd()
if not (PROJECT_ROOT / "src").exists():
    PROJECT_ROOT = PROJECT_ROOT.parent

if str(PROJECT_ROOT) not in sys.path:
    sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd

from src.models import GNNConfig, build_pooled_dataset

pd.set_option("display.max_columns", 200)
```

## Configuration

```python
cfg = GNNConfig(
    model_type="GraphSAGE",
    hidden_dims=(256, 64),
    dropout=0.3,
    lr=0.01,
    epochs=100,
    aggregation="mean",
    device="cpu",
)

TARGET_COL = "systemic_risk_label"
INCLUDE_RAW_FEATURES = False

OUTPUT_DATASET = PROJECT_ROOT / "outputs" / "embeddings" / "graphsage_srisk_dataset.parquet"
OUTPUT_DATASET
```

## Build Embedding Dataset

For each quarter, GraphSAGE is trained on that quarter's graph only. The resulting embeddings are merged with the regression target `systemic_risk_label` and optionally with the raw node features.

```python
pooled_df = build_pooled_dataset(
    config=cfg,
    years=range(2016, 2024),
    quarters=(1, 2, 3, 4),
    target_col=TARGET_COL,
    include_raw_features=INCLUDE_RAW_FEATURES,
    output_path=OUTPUT_DATASET,
)

pooled_df.shape
```

```python
pooled_df.head()
```

```python
embedding_cols = [c for c in pooled_df.columns if c.startswith("emb_")]
meta_cols = ["bank_id", "year", "quarter", "period", TARGET_COL]
meta_cols + embedding_cols[:5], len(embedding_cols)
```

```python
pooled_df.groupby(["year", "quarter"])[TARGET_COL].agg(["count", "mean", "max"]).head(12)
```

## Exported Dataset

The saved dataset contains:
- metadata: `bank_id`, `year`, `quarter`, `period`
- target: `systemic_risk_label`
- embeddings: `emb_0`, `emb_1`, ...
- optional raw features from the node table
