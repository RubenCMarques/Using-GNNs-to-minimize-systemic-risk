# Dataset 1 - per-p embeddings (graphsage_v1_32)

Re-runs the **exact same** GraphSAGE extraction used in notebook 08 (`graphsage_v1_32`, temporal/warm-started across the 32 quarters) once for **each p**, pooling that p's targets. Same network, same features, same config - only the target (cascade size at that default threshold) changes at the end. Outputs one parquet per p next to the p0 baseline `graphsage_v1_32_srisk_dataset.parquet`.

```python
import sys
from pathlib import Path

def find_project_root(start=None):
    start = Path(start or Path.cwd()).resolve()
    for path in [start, *start.parents]:
        if (path / 'src').exists() and (path / 'requirements.txt').exists():
            return path
    raise FileNotFoundError('Project root not found.')

PROJECT_ROOT = find_project_root()
sys.path.insert(0, str(PROJECT_ROOT))

import pandas as pd

from src.models.embeddings import GNNConfig
from src.models.embedding_pipeline import build_pooled_dataset

EMB_ROOT = PROJECT_ROOT / 'src' / 'data' / 'embeddings'
TARGETS_DIR = PROJECT_ROOT / 'src' / 'datasets' / 'dataset_1' / 'targets'

def emb_path(name):
    return EMB_ROOT / 'dataset_1' / 'feature_based' / name

TARGET_COL = 'log_systemic_risk_label'
YEARS      = range(2016, 2024)
QUARTERS   = (1, 2, 3, 4)
P_LIST = ['p5', 'p10', 'p15', 'p20', 'p25', 'p30', 'p35', 'p40']

# Same config as the selected graphsage_v1_32 embedding (see notebooks 03/08).
cfg_graphsage_v1_32 = GNNConfig(hidden_dims=(256, 32), dropout=0.3, lr=0.01,
                                epochs=100, aggregation='mean', device='cpu')
```

```python
for p in P_LIST:
    out = emb_path(f'graphsage_v1_32_{p}_dataset.parquet')
    df = build_pooled_dataset(
        config=cfg_graphsage_v1_32,
        years=YEARS,
        quarters=QUARTERS,
        target_col=TARGET_COL,
        include_raw_features=False,
        target_dir=TARGETS_DIR / p,     # this p's per-quarter targets
        output_path=out,
    )
    emb_cols = [c for c in df.columns if c.startswith('emb_')]
    print(f'{p:4s}  shape={df.shape}  emb_cols={len(emb_cols)}  -> {out.name}')
```

```python
files = sorted(emb_path('').glob('graphsage_v1_32_p*_dataset.parquet'))
for f in files:
    print(f'{f.name:45s}  shape={pd.read_parquet(f).shape}')
```
